```python
import pandas as pd
import numpy as np
from google.colab import files

print("Libraries imported successfully.")

print("Please upload your Excel dataset:")

uploaded = files.upload()

file_name = next(iter(uploaded))

print(f"\nFile uploaded successfully: {file_name}")


df = pd.read_excel(file_name)

print("\n" + "="*60)
print("DATASET LOADED")
print("="*60)

print(f"Number of rows    : {df.shape[0]}")
print(f"Number of columns : {df.shape[1]}")

print("\nFirst 5 rows:")
display(df.head())


print("\n" + "="*60)
print("INITIAL DATA INFORMATION")
print("="*60)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()

print("\n" + "="*60)
print("MISSING VALUES ANALYSIS")
print("="*60)

missing_count = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_count,
    "Missing Percentage": missing_percentage.round(2)
})

missing_summary = missing_summary[
    missing_summary["Missing Values"] > 0
].sort_values("Missing Values", ascending=False)

if missing_summary.empty:
    print("No missing values found.")
else:
    print("Missing values found:")
    display(missing_summary)




if "CouponCode" in df.columns:
    df["CouponCode"] = df["CouponCode"].fillna("No Coupon")

print("\n" + "="*60)
print("DUPLICATE ANALYSIS")
print("="*60)

duplicates_before = df.duplicated().sum()

print(f"Duplicate rows found: {duplicates_before}")

if duplicates_before > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print(f"Duplicate rows removed: {duplicates_before}")
else:
    print("No duplicate rows found.")


df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
)

text_columns = df.select_dtypes(include=["object"]).columns

for column in text_columns:
    df[column] = df[column].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("\nText columns cleaned successfully.")


if "Date" in df.columns:
    df["Date"] = pd.to_datetime(
        df["Date"],
        errors="coerce"
    )

    print("Date column converted to datetime format.")

numeric_columns = [
    "Quantity",
    "UnitPrice",
    "ItemsInCart",
    "TotalPrice"
]

for column in numeric_columns:
    if column in df.columns:
        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )

print("Numeric columns converted successfully.")



standardize_columns = [
    "Product",
    "PaymentMethod",
    "OrderStatus",
    "CouponCode",
    "ReferralSource"
]

for column in standardize_columns:
    if column in df.columns:
        df[column] = df[column].apply(
            lambda x: " ".join(x.split()) if isinstance(x, str) else x
        )

print("\n" + "="*60)
print("MISSING VALUES AFTER CLEANING")
print("="*60)

remaining_missing = df.isnull().sum()

remaining_missing = remaining_missing[
    remaining_missing > 0
]

if remaining_missing.empty:
    print("No missing values remain.")
else:
    print("Remaining missing values:")
    display(remaining_missing)


print("\n" + "="*60)
print("DUPLICATES AFTER CLEANING")
print("="*60)

duplicates_after = df.duplicated().sum()

print(f"Remaining duplicate rows: {duplicates_after}")


print("\n" + "="*60)
print("FINAL DATA TYPES")
print("="*60)

display(
    pd.DataFrame({
        "Column": df.columns,
        "Data Type": df.dtypes.astype(str).values
    })
)


print("\n" + "="*60)
print("DATA QUALITY CHECKS")
print("="*60)

if "Quantity" in df.columns:
    negative_quantity = (df["Quantity"] < 0).sum()
    print(f"Negative Quantity values: {negative_quantity}")

if "UnitPrice" in df.columns:
    negative_price = (df["UnitPrice"] < 0).sum()
    print(f"Negative UnitPrice values: {negative_price}")


if "TotalPrice" in df.columns:
    negative_total = (df["TotalPrice"] < 0).sum()
    print(f"Negative TotalPrice values: {negative_total}")

if "Date" in df.columns:
    invalid_dates = df["Date"].isnull().sum()
    print(f"Invalid/Missing Date values: {invalid_dates}")

if all(col in df.columns for col in ["Quantity", "UnitPrice", "TotalPrice"]):

    calculated_total = (
        df["Quantity"] * df["UnitPrice"]
    ).round(2)

    price_mismatch = (
        calculated_total != df["TotalPrice"].round(2)
    ).sum()

    print(f"TotalPrice calculation mismatches: {price_mismatch}")

print("\n" + "="*60)
print("FINAL CLEAN DATASET SUMMARY")
print("="*60)

print(f"Rows before cleaning     : 1200")
print(f"Rows after cleaning      : {len(df)}")
print(f"Columns                  : {len(df.columns)}")
print(f"Duplicate rows remaining : {df.duplicated().sum()}")
print(f"Missing values remaining : {df.isnull().sum().sum()}")

print("\nFinal dataset preview:")
display(df.head())

output_file = "Cleaned_Dataset.xlsx"

df.to_excel(
    output_file,
    index=False
)

print("\n" + "="*60)
print("CLEANING COMPLETED")
print("="*60)

print(f"Cleaned dataset saved as: {output_file}")

files.download(output_file)
```
