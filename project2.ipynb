```python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files
from IPython.display import display

print("✅ Libraries imported successfully")



print("📂 Upload the CLEANED dataset from Project 1")
print("Example: Cleaned_Dataset.xlsx")

uploaded = files.upload()

file_name = next(iter(uploaded))

print(f"\n✅ File uploaded: {file_name}")



df = pd.read_excel(file_name)

# Make a copy for EDA
eda = df.copy()

print("\n" + "="*70)
print("PROJECT 2 - EDA")
print("="*70)

print(f"Rows    : {eda.shape[0]:,}")
print(f"Columns : {eda.shape[1]}")

print("\n📌 First 5 rows:")
display(eda.head())



print("\n" + "="*70)
print("1. DATA STRUCTURE")
print("="*70)

print("\nColumn names:")
print(list(eda.columns))

print("\nData types:")
display(
    pd.DataFrame({
        "Column": eda.columns,
        "Data Type": eda.dtypes.astype(str),
        "Non-Null": eda.notnull().sum(),
        "Unique": eda.nunique()
    }).reset_index(drop=True)
)



print("\n" + "="*70)
print("2. DATA QUALITY CHECK")
print("="*70)

missing = eda.isnull().sum()
duplicates = eda.duplicated().sum()

print(f"Total missing values    : {missing.sum():,}")
print(f"Total duplicate rows    : {duplicates:,}")

if missing.sum() > 0:
    print("\nMissing values by column:")
    display(
        missing[missing > 0]
        .to_frame("Missing Values")
    )
else:
    print("✅ No missing values found.")

if duplicates == 0:
    print("✅ No duplicate rows found.")
else:
    print(f"⚠️ {duplicates} duplicate rows found.")


numeric_columns = eda.select_dtypes(
    include=np.number
).columns.tolist()

categorical_columns = eda.select_dtypes(
    include=["object", "category"]
).columns.tolist()

date_columns = eda.select_dtypes(
    include=["datetime64[ns]", "datetime64[ns, UTC]"]
).columns.tolist()

print("\n" + "="*70)
print("3. VARIABLE TYPES")
print("="*70)

print("Numeric columns:")
print(numeric_columns)

print("\nCategorical/Text columns:")
print(categorical_columns)

print("\nDate columns:")
print(date_columns)



print("\n" + "="*70)
print("4. DESCRIPTIVE STATISTICS")
print("="*70)

if len(numeric_columns) > 0:

    descriptive_stats = pd.DataFrame({
        "Count": eda[numeric_columns].count(),
        "Mean": eda[numeric_columns].mean(),
        "Median": eda[numeric_columns].median(),
        "Std": eda[numeric_columns].std(),
        "Min": eda[numeric_columns].min(),
        "Q1": eda[numeric_columns].quantile(0.25),
        "Q3": eda[numeric_columns].quantile(0.75),
        "Max": eda[numeric_columns].max(),
        "Range": (
            eda[numeric_columns].max()
            - eda[numeric_columns].min()
        )
    })

    display(
        descriptive_stats.round(2)
    )

else:
    print("No numeric columns available.")


print("\n" + "="*70)
print("5. NUMERICAL DISTRIBUTIONS")
print("="*70)

for column in numeric_columns:

    data = eda[column].dropna()

    if len(data) == 0:
        continue

    plt.figure(figsize=(10, 5))

    plt.hist(
        data,
        bins=30,
        edgecolor="black"
    )

    plt.title(
        f"Distribution of {column}",
        fontsize=14,
        fontweight="bold"
    )

    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.grid(
        axis="y",
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()



print("\n" + "="*70)
print("6. OUTLIER ANALYSIS")
print("="*70)

outlier_results = []

for column in numeric_columns:

    data = eda[column].dropna()

    if len(data) == 0:
        continue

    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data < lower_bound) |
        (data > upper_bound)
    ]

    outlier_results.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers),
        "Outlier %": len(outliers) / len(data) * 100
    })

    # Boxplot
    plt.figure(figsize=(10, 4))

    plt.boxplot(
        data,
        vert=False
    )

    plt.title(
        f"Boxplot - {column}",
        fontsize=14,
        fontweight="bold"
    )

    plt.xlabel(column)

    plt.grid(
        axis="x",
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()


outlier_summary = pd.DataFrame(
    outlier_results
)

if not outlier_summary.empty:
    display(
        outlier_summary.round(2)
    )


print("\n" + "="*70)
print("7. CATEGORICAL ANALYSIS")
print("="*70)

for column in categorical_columns:

    # Avoid displaying extremely high-cardinality columns
    if eda[column].nunique() <= 20:

        print(f"\n📊 {column}")

        counts = eda[column].value_counts(
            dropna=False
        )

        percentages = (
            eda[column]
            .value_counts(
                normalize=True,
                dropna=False
            )
            .mul(100)
            .round(2)
        )

        summary = pd.DataFrame({
            "Count": counts,
            "Percentage": percentages
        })

        display(summary)

        # Plot
        plt.figure(figsize=(10, 5))

        plt.bar(
            summary.index.astype(str),
            summary["Count"]
        )

        plt.title(
            f"Distribution of {column}",
            fontsize=14,
            fontweight="bold"
        )

        plt.xlabel(column)
        plt.ylabel("Count")

        plt.xticks(
            rotation=35,
            ha="right"
        )

        plt.tight_layout()
        plt.show()



if "Product" in eda.columns:

    print("\n" + "="*70)
    print("8. PRODUCT ANALYSIS")
    print("="*70)

    product_summary = eda.groupby("Product").agg(
        Orders=("OrderID", "count"),
        Quantity=("Quantity", "sum"),
        Revenue=("TotalPrice", "sum"),
        Average_Order_Value=("TotalPrice", "mean"),
        Average_Unit_Price=("UnitPrice", "mean")
    ).sort_values(
        "Revenue",
        ascending=False
    )

    display(
        product_summary.round(2)
    )

    # Revenue by Product
    plt.figure(figsize=(11, 6))

    plt.bar(
        product_summary.index.astype(str),
        product_summary["Revenue"]
    )

    plt.title(
        "Revenue by Product",
        fontsize=15,
        fontweight="bold"
    )

    plt.xlabel("Product")
    plt.ylabel("Total Revenue")

    plt.xticks(
        rotation=40,
        ha="right"
    )

    plt.tight_layout()
    plt.show()

    # Quantity by Product
    plt.figure(figsize=(11, 6))

    plt.bar(
        product_summary.index.astype(str),
        product_summary["Quantity"]
    )

    plt.title(
        "Quantity Sold by Product",
        fontsize=15,
        fontweight="bold"
    )

    plt.xlabel("Product")
    plt.ylabel("Quantity Sold")

    plt.xticks(
        rotation=40,
        ha="right"
    )

    plt.tight_layout()
    plt.show()



if "OrderStatus" in eda.columns:

    print("\n" + "="*70)
    print("9. ORDER STATUS ANALYSIS")
    print("="*70)

    status_summary = eda.groupby(
        "OrderStatus"
    ).agg(
        Orders=("OrderID", "count"),
        Revenue=("TotalPrice", "sum"),
        Average_Order_Value=("TotalPrice", "mean")
    ).sort_values(
        "Orders",
        ascending=False
    )

    display(
        status_summary.round(2)
    )

    plt.figure(figsize=(10, 5))

    plt.bar(
        status_summary.index.astype(str),
        status_summary["Orders"]
    )

    plt.title(
        "Order Status Distribution",
        fontsize=15,
        fontweight="bold"
    )

    plt.xlabel("Order Status")
    plt.ylabel("Number of Orders")

    plt.xticks(
        rotation=30
    )

    plt.tight_layout()
    plt.show()



if "PaymentMethod" in eda.columns:

    print("\n" + "="*70)
    print("10. PAYMENT METHOD ANALYSIS")
    print("="*70)

    payment_summary = eda.groupby(
        "PaymentMethod"
    ).agg(
        Orders=("OrderID", "count"),
        Revenue=("TotalPrice", "sum"),
        Average_Order_Value=("TotalPrice", "mean")
    ).sort_values(
        "Orders",
        ascending=False
    )

    display(
        payment_summary.round(2)
    )

    plt.figure(figsize=(10, 5))

    plt.bar(
        payment_summary.index.astype(str),
        payment_summary["Orders"]
    )

    plt.title(
        "Orders by Payment Method",
        fontsize=15,
        fontweight="bold"
    )

    plt.xlabel("Payment Method")
    plt.ylabel("Number of Orders")

    plt.xticks(
        rotation=30
    )

    plt.tight_layout()
    plt.show()


if "ReferralSource" in eda.columns:

    print("\n" + "="*70)
    print("11. REFERRAL SOURCE ANALYSIS")
    print("="*70)

    referral_summary = eda.groupby(
        "ReferralSource"
    ).agg(
        Orders=("OrderID", "count"),
        Revenue=("TotalPrice", "sum"),
        Average_Order_Value=("TotalPrice", "mean")
    ).sort_values(
        "Revenue",
        ascending=False
    )

    display(
        referral_summary.round(2)
    )

    plt.figure(figsize=(10, 5))

    plt.bar(
        referral_summary.index.astype(str),
        referral_summary["Revenue"]
    )

    plt.title(
        "Revenue by Referral Source",
        fontsize=15,
        fontweight="bold"
    )

    plt.xlabel("Referral Source")
    plt.ylabel("Revenue")

    plt.xticks(
        rotation=30
    )

    plt.tight_layout()
    plt.show()



if "CouponCode" in eda.columns:

    print("\n" + "="*70)
    print("12. COUPON ANALYSIS")
    print("="*70)

    coupon_summary = eda.groupby(
        "CouponCode"
    ).agg(
        Orders=("OrderID", "count"),
        Revenue=("TotalPrice", "sum"),
        Average_Order_Value=("TotalPrice", "mean")
    ).sort_values(
        "Orders",
        ascending=False
    )

    display(
        coupon_summary.round(2)
    )

    plt.figure(figsize=(10, 5))

    plt.bar(
        coupon_summary.index.astype(str),
        coupon_summary["Orders"]
    )

    plt.title(
        "Orders by Coupon",
        fontsize=15,
        fontweight="bold"
    )

    plt.xlabel("Coupon")
    plt.ylabel("Number of Orders")

    plt.xticks(
        rotation=30
    )

    plt.tight_layout()
    plt.show()


if "Date" in eda.columns:

    print("\n" + "="*70)
    print("13. TIME TREND ANALYSIS")
    print("="*70)

    eda["Date"] = pd.to_datetime(
        eda["Date"],
        errors="coerce"
    )

    # Remove invalid dates for time analysis
    time_df = eda.dropna(
        subset=["Date"]
    ).copy()

    # Monthly
    monthly = (
        time_df
        .set_index("Date")
        .resample("ME")
        .agg(
            Orders=("OrderID", "count"),
            Revenue=("TotalPrice", "sum"),
            Quantity=("Quantity", "sum")
        )
    )

    print("\nMonthly Performance:")
    display(
        monthly.round(2)
    )

    # Monthly Revenue
    plt.figure(figsize=(13, 5))

    plt.plot(
        monthly.index,
        monthly["Revenue"],
        marker="o"
    )

    plt.title(
        "Monthly Revenue Trend",
        fontsize=15,
        fontweight="bold"
    )

    plt.xlabel("Month")
    plt.ylabel("Revenue")

    plt.xticks(
        rotation=45
    )

    plt.grid(
        True,
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()

    # Monthly Orders
    plt.figure(figsize=(13, 5))

    plt.plot(
        monthly.index,
        monthly["Orders"],
        marker="o"
    )

    plt.title(
        "Monthly Order Volume Trend",
        fontsize=15,
        fontweight="bold"
    )

    plt.xlabel("Month")
    plt.ylabel("Orders")

    plt.xticks(
        rotation=45
    )

    plt.grid(
        True,
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()



if "Date" in eda.columns:

    time_df["Day"] = time_df["Date"].dt.day_name()

    day_order = [
        "Monday",
        "Tuesday",
        "Wednesday",
        "Thursday",
        "Friday",
        "Saturday",
        "Sunday"
    ]

    day_summary = time_df.groupby(
        "Day"
    ).agg(
        Orders=("OrderID", "count"),
        Revenue=("TotalPrice", "sum")
    ).reindex(day_order)

    print("\n" + "="*70)
    print("14. DAY OF WEEK ANALYSIS")
    print("="*70)

    display(
        day_summary.round(2)
    )

    plt.figure(figsize=(10, 5))

    plt.bar(
        day_summary.index,
        day_summary["Orders"]
    )

    plt.title(
        "Orders by Day of Week",
        fontsize=15,
        fontweight="bold"
    )

    plt.xlabel("Day")
    plt.ylabel("Orders")

    plt.xticks(
        rotation=30
    )

    plt.tight_layout()
    plt.show()


print("\n" + "="*70)
print("15. CORRELATION ANALYSIS")
print("="*70)

if len(numeric_columns) >= 2:

    correlation = eda[numeric_columns].corr()

    print("\nCorrelation Matrix:")
    display(
        correlation.round(2)
    )

    # Correlation matrix visualization
    plt.figure(figsize=(10, 8))

    plt.imshow(
        correlation,
        aspect="auto"
    )

    plt.colorbar(
        label="Correlation"
    )

    plt.xticks(
        range(len(correlation.columns)),
        correlation.columns,
        rotation=45,
        ha="right"
    )

    plt.yticks(
        range(len(correlation.columns)),
        correlation.columns
    )

    plt.title(
        "Correlation Matrix",
        fontsize=15,
        fontweight="bold"
    )

    plt.tight_layout()
    plt.show()

else:
    print("Not enough numerical variables for correlation analysis.")



print("\n" + "="*70)
print("16. KEY BUSINESS KPIs")
print("="*70)

kpis = {}

if "OrderID" in eda.columns:
    kpis["Total Orders"] = eda["OrderID"].nunique()

if "CustomerID" in eda.columns:
    kpis["Unique Customers"] = eda["CustomerID"].nunique()

if "Product" in eda.columns:
    kpis["Unique Products"] = eda["Product"].nunique()

if "TotalPrice" in eda.columns:
    kpis["Total Revenue"] = eda["TotalPrice"].sum()
    kpis["Average Order Value"] = eda["TotalPrice"].mean()
    kpis["Median Order Value"] = eda["TotalPrice"].median()

if "Quantity" in eda.columns:
    kpis["Total Quantity Sold"] = eda["Quantity"].sum()
    kpis["Average Quantity per Order"] = eda["Quantity"].mean()

kpi_table = pd.DataFrame({
    "KPI": list(kpis.keys()),
    "Value": list(kpis.values())
})

display(
    kpi_table.round(2)
)



print("\n" + "="*70)
print("17. KEY OBSERVATIONS")
print("="*70)

observations = []


# Product insight
if "Product" in eda.columns and "TotalPrice" in eda.columns:

    product_revenue = (
        eda.groupby("Product")["TotalPrice"]
        .sum()
        .sort_values(ascending=False)
    )

    top_product = product_revenue.index[0]
    top_product_value = product_revenue.iloc[0]

    observations.append(
        f"• The highest-revenue product is "
        f"'{top_product}', generating "
        f"{top_product_value:,.2f} in total revenue."
    )


# Order status insight
if "OrderStatus" in eda.columns:

    status_counts = eda["OrderStatus"].value_counts()

    most_common_status = status_counts.index[0]
    most_common_status_count = status_counts.iloc[0]

    observations.append(
        f"• The most common order status is "
        f"'{most_common_status}', representing "
        f"{most_common_status_count:,} orders."
    )


# Payment insight
if "PaymentMethod" in eda.columns:

    payment_counts = eda["PaymentMethod"].value_counts()

    most_used_payment = payment_counts.index[0]
    payment_count = payment_counts.iloc[0]

    observations.append(
        f"• The most frequently used payment method is "
        f"'{most_used_payment}', used in "
        f"{payment_count:,} orders."
    )


# Referral insight
if "ReferralSource" in eda.columns and "TotalPrice" in eda.columns:

    referral_revenue = (
        eda.groupby("ReferralSource")["TotalPrice"]
        .sum()
        .sort_values(ascending=False)
    )

    top_referral = referral_revenue.index[0]
    top_referral_value = referral_revenue.iloc[0]

    observations.append(
        f"• The referral source generating the highest revenue is "
        f"'{top_referral}', with "
        f"{top_referral_value:,.2f} in revenue."
    )


# Monthly insight
if "Date" in eda.columns and "TotalPrice" in eda.columns:

    if "monthly" in locals() and not monthly.empty:

        highest_month = monthly["Revenue"].idxmax()
        highest_month_value = monthly["Revenue"].max()

        lowest_month = monthly["Revenue"].idxmin()
        lowest_month_value = monthly["Revenue"].min()

        observations.append(
            f"• The highest-revenue month is "
            f"{highest_month.strftime('%B %Y')}, generating "
            f"{highest_month_value:,.2f}."
        )

        observations.append(
            f"• The lowest-revenue month is "
            f"{lowest_month.strftime('%B %Y')}, generating "
            f"{lowest_month_value:,.2f}."
        )


# Outlier insight
if not outlier_summary.empty:

    highest_outlier = (
        outlier_summary
        .sort_values(
            "Outlier Count",
            ascending=False
        )
        .iloc[0]
    )

    observations.append(
        f"• '{highest_outlier['Column']}' has the highest number "
        f"of potential outliers: "
        f"{int(highest_outlier['Outlier Count'])} observations "
        f"({highest_outlier['Outlier %']:.2f}%)."
    )


# Print observations
for observation in observations:
    print(observation)




print("\n" + "="*70)
print("PROJECT 2 - FINAL SUMMARY")
print("="*70)

print(f"""
Dataset analyzed:
• Rows: {len(eda):,}
• Columns: {len(eda.columns)}

EDA components completed:
✓ Basic statistics
✓ Mean / Median / Count
✓ Data distributions
✓ Outlier detection
✓ Categorical analysis
✓ Product analysis
✓ Order status analysis
✓ Payment method analysis
✓ Referral source analysis
✓ Coupon analysis
✓ Monthly trends
✓ Day-of-week trends
✓ Correlation analysis
✓ Business KPIs
✓ Key observations

The cleaned dataset has been successfully analyzed.
""")


output_file = "DecodeLabs_Project_2_EDA_Results.xlsx"

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    # Raw EDA dataset
    eda.to_excel(
        writer,
        sheet_name="EDA_Data",
        index=False
    )

    # Column information
    column_info = pd.DataFrame({
        "Column": eda.columns,
        "Data Type": eda.dtypes.astype(str).values,
        "Non-Null": eda.notnull().sum().values,
        "Missing": eda.isnull().sum().values,
        "Unique": eda.nunique().values
    })

    column_info.to_excel(
        writer,
        sheet_name="Column_Info",
        index=False
    )

    # Statistics
    if len(numeric_columns) > 0:
        descriptive_stats.to_excel(
            writer,
            sheet_name="Statistics"
        )

    # Outliers
    if not outlier_summary.empty:
        outlier_summary.to_excel(
            writer,
            sheet_name="Outliers",
            index=False
        )

    # KPIs
    kpi_table.to_excel(
        writer,
        sheet_name="KPIs",
        index=False
    )

    # Correlation
    if len(numeric_columns) >= 2:
        correlation.to_excel(
            writer,
            sheet_name="Correlation"
        )

    # Product
    if "Product" in locals():
        product_summary.to_excel(
            writer,
            sheet_name="Products"
        )

    # Status
    if "status_summary" in locals():
        status_summary.to_excel(
            writer,
            sheet_name="Order_Status"
        )

    # Payment
    if "payment_summary" in locals():
        payment_summary.to_excel(
            writer,
            sheet_name="Payment"
        )

    # Referral
    if "referral_summary" in locals():
        referral_summary.to_excel(
            writer,
            sheet_name="Referral"
        )

    # Coupon
    if "coupon_summary" in locals():
        coupon_summary.to_excel(
            writer,
            sheet_name="Coupons"
        )

    # Monthly
    if "monthly" in locals():
        monthly.to_excel(
            writer,
            sheet_name="Monthly_Trend"
        )

    # Day
    if "day_summary" in locals():
        day_summary.to_excel(
            writer,
            sheet_name="Day_Analysis"
        )


print("\n" + "="*70)
print("✅ EDA RESULTS EXPORTED")
print("="*70)

print(f"File created: {output_file}")



files.download(output_file)

print("\n🎉 PROJECT 2 COMPLETED SUCCESSFULLY!")
print("Your EDA results file is ready.")
```
